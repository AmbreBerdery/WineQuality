## 1. Données et variables retenues

```python
df = make_binary_target(clean_data(load_wine_data()))
X_train, X_test, y_train, y_test = split_data(df)
features = load_selected_features()

print("Train :", X_train.shape, "| Test :", X_test.shape)
print("Variables retenues (notebook 02) :", features)
```

## 2. Régression logistique

Pourquoi ce modèle ?
- C'est un modèle linéaire : il fait une somme pondérée des variables, puis la transforme en probabilité avec la fonction sigmoïde. La frontière entre bons et moins bons vins est donc une droite (un hyperplan en plusieurs dimensions).
- Il est simple et stable, donc peu sujet au surapprentissage. Il est aussi interprétable : le signe d'un coefficient indique si la variable augmente ou diminue la probabilité d'un bon vin.
- Sa limite : si la vraie relation n'est pas linéaire, il manque de précision (biais).
- Il convient à nos données : les variables sont continues et standardisées par le pipeline, ce qui est nécessaire car la régularisation dépend de l'échelle des variables. `class_weight="balanced"` compense le déséquilibre des classes.
- Hyperparamètres optimisés : `C` (plus il est petit, plus le modèle est contraint) et la pénalité (L1, L2 ou un mélange des deux, via `l1_ratio`).

### 2.1 Modèle par défaut (C = 1, L2)

```python
logreg_base = logistic_regression(features)
pd.Series(train_vs_cv_scores(logreg_base, X_train, y_train)).round(3)
```

### 2.2 Grid Search sur `C` et la pénalité

```python
logreg_search = run_grid_search(logistic_regression(features), LOGREG_GRID, X_train, y_train)

print("Meilleurs hyperparamètres :", logreg_search.best_params_)
print("F1 moyen en validation croisée :", round(logreg_search.best_score_, 3))
```

```python
logreg_cv = pd.DataFrame(logreg_search.cv_results_)
pivot = logreg_cv.pivot_table(index="param_model__C", columns="param_model__l1_ratio", values="mean_test_score")
pivot.index = [f"{c:.3g}" for c in pivot.index]

plt.figure(figsize=(6, 6))
sns.heatmap(pivot, annot=True, fmt=".3f", cmap="viridis")
plt.xlabel("l1_ratio (0 = L2, 0.5 = elastic-net, 1 = L1)")
plt.ylabel("C")
plt.title("Régression logistique : F1 moyen en validation croisée")
plt.tight_layout()
plt.show()
```

```python
coefs = pd.Series(
    logreg_search.best_estimator_.named_steps["model"].coef_[0], index=features
).sort_values()

coefs.plot.barh(figsize=(7, 3), color=["tab:red" if c < 0 else "tab:green" for c in coefs])
plt.axvline(0, color="black", lw=0.8)
plt.title("Coefficients de la régression logistique (variables standardisées)")
plt.tight_layout()
plt.show()
coefs.round(3)
```

## 3. Arbre de décision

Pourquoi ce modèle ?
- C'est un modèle non linéaire : il enchaîne des questions du type « alcool > 11 % ? ». La frontière est en escalier, et l'arbre peut capturer des interactions entre variables.
- Il n'a pas besoin de standardisation et il est peu sensible aux valeurs aberrantes et aux distributions asymétriques, car seul l'ordre des valeurs compte.
- Sa limite : sans contrainte, il grandit jusqu'à apprendre le train par cœur. C'est le surapprentissage (variance élevée).
- Hyperparamètres optimisés : `max_depth` (profondeur maximale) et `min_samples_leaf` (nombre minimal d'exemples par feuille), qui limitent sa complexité.

### 3.1 Arbre non contraint : mise en évidence du surapprentissage

```python
tree_base = decision_tree(features)
pd.Series(train_vs_cv_scores(tree_base, X_train, y_train)).round(3)
```

Analyse. F1 = 1,00 sur le train mais 0,39 en validation : l'arbre non contraint apprend le train par cœur, bruit compris. C'est l'illustration typique d'une variance élevée, et il fait moins bien que la régression logistique (0,525).

### 3.2 Grid Search sur `max_depth` et `min_samples_leaf`

```python
tree_search = run_grid_search(decision_tree(features), TREE_GRID, X_train, y_train)

print("Meilleurs hyperparamètres :", tree_search.best_params_)
print("F1 moyen en validation croisée :", round(tree_search.best_score_, 3))
```

```python
tree_cv = pd.DataFrame(tree_search.cv_results_)
tree_cv["param_model__max_depth"] = tree_cv["param_model__max_depth"].fillna("None").astype(str)
depth_order = [str(d) for d in TREE_GRID["model__max_depth"]]

fig, axes = plt.subplots(1, 2, figsize=(14, 5))
for ax, score, title in [(axes[0], "mean_train_score", "Entraînement"), (axes[1], "mean_test_score", "Validation croisée")]:
    pivot = tree_cv.pivot_table(index="param_model__max_depth", columns="param_model__min_samples_leaf", values=score)
    sns.heatmap(pivot.loc[depth_order], annot=True, fmt=".2f", cmap="viridis", ax=ax, vmin=0.3, vmax=1)
    ax.set_xlabel("min_samples_leaf")
    ax.set_ylabel("max_depth")
    ax.set_title(f"Arbre : F1 moyen ({title})")
plt.tight_layout()
plt.show()
```

```python
from sklearn.tree import plot_tree

best_tree = tree_search.best_estimator_.named_steps["model"]

plt.figure(figsize=(18, 7))
plot_tree(best_tree, feature_names=features, class_names=["moins bon", "bon"],
          filled=True, rounded=True, max_depth=3, fontsize=9)
plt.title("Arbre optimisé (3 premiers niveaux, seuils sur variables standardisées)")
plt.show()

pd.Series(best_tree.feature_importances_, index=features).sort_values(ascending=False).round(3)
```

## 4. Naive Bayes

### 4.1 Vérification des hypothèses sur le train

```python
from src.models import naive_bayes

print("Corrélations entre les variables retenues (hypothèse d'indépendance) :")
display(X_train[features].corr().round(2))

print("Coefficient d'asymétrie par classe (hypothèse gaussienne, 0 pour une loi normale) :")
X_train[features].groupby(y_train).skew().round(2)
```

**Analyse.** `density` est fortement corrélée à `alcohol` et à `residual sugar` : l'hypothèse d'indépendance est violée, et le modèle compte en partie deux fois la même information. Plusieurs variables sont asymétriques dans au moins une classe, ce qui s'écarte de l'hypothèse gaussienne.

### 4.2 Modèle standard et variante à priors égaux

```python
nb_variants = {
    "standard, 4 variables": naive_bayes(features),
    "priors égaux, 4 variables": naive_bayes(features, priors=[0.5, 0.5]),
    "priors égaux, 12 variables": naive_bayes(None, priors=[0.5, 0.5]),
}
pd.DataFrame(
    {name: train_vs_cv_scores(model, X_train, y_train) for name, model in nb_variants.items()}
).T.round(3)
```

**Analyse.** Les priors égaux améliorent le F1 en validation croisée (0,488 à 0,498) : le modèle prédit plus souvent « bon », ce qui augmente le rappel. Avec les 12 variables, le F1 baisse (0,476) : les variables supplémentaires, corrélées et asymétriques, aggravent la violation des hypothèses. L'écart entre le train et la validation est nul : Naive Bayes ne surapprend pas, il est limité par son biais.

**Modèle retenu :** priors égaux, 4 variables retenues au notebook 02.

```python
nb_final = naive_bayes(features, priors=[0.5, 0.5]).fit(X_train, y_train)
pd.Series(train_vs_cv_scores(nb_final, X_train, y_train)).round(3)
```
